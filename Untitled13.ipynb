```python
## Data Preprocessing
```

```python
data = pd.get_dummies(data, columns=['Gender'], drop_first=True)
```

```python
mapping = {'Low':1, 'Middle':2, 'High':3}
data['Income_Level'] = data['Income_Level'].map(mapping)
```

```python
data['Marital_Status'].unique()
```

```python
mapping = {'Married': 1, 'Single': 2, 'Widowed': 3, 'Divorced': 4}
data['Marital_Status'] = data['Marital_Status'].map(mapping)
```

```python
data['Education_Level'].value_counts()
```

```python
mapping = {"Bachelor's": 1, 'High School': 2, "Master's": 3}
data['Education_Level'] = data['Education_Level'].map(mapping)
```

```python
data['Occupation'].value_counts()
```

```python
mapping = {"High": 1, 'Middle': 2}
data['Occupation'] = data['Occupation'].map(mapping)
```

```python
data.head()
```

```python
bool_cols = data.select_dtypes(include='bool').columns
data[bool_cols] = data[bool_cols].astype(int)
```

```python
data.head()
```

```python
data['Purchase_Amount'].info()
```

```python
value = data['Purchase_Amount'][0]
value
```

```python
val = value.replace('$', '').replace(' ', '')
val = float(val)
val
```

```python
data['Purchase_Amount'] = data['Purchase_Amount'].apply(lambda x: float(x.replace('$', '').replace(' ', '')))
```

```python
data['Purchase_Amount'].info()
```

```python
data['Month'] = pd.to_datetime(data['Time_of_Purchase']).dt.month
data['Day'] = pd.to_datetime(data['Time_of_Purchase']).dt.day
```

```python
data.drop(columns=['Customer_ID'], inplace=True)
data.drop(columns=['Location'], inplace=True)
data.drop(columns=['Time_of_Purchase'], inplace=True)
```

```python
data.head()
```

```python
data['Purchase_Category'].unique()
```

```python
data['Purchase_Category'] = data['Purchase_Category'].str.strip()
data['Purchase_Category'] = data['Purchase_Category'].replace({
    'Packages)': 'Packages',
    'Travel & Leisure (Flights': 'Travel & Leisure'
})
```

```python
data['Purchase_Category'].nunique()
data = pd.get_dummies(data, columns=['Purchase_Category'], drop_first=True, dtype=int)
```

```python
data.head()
```

```python
data['Purchase_Channel'].unique()
```

```python
mapping = {"Mixed": 1, 'In-Store': 2, 'Online': 3}
data['Purchase_Channel'] = data['Purchase_Channel'].map(mapping)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
X = data.drop(columns=['Customer_Loyalty_Program_Member'])
y = data['Customer_Loyalty_Program_Member']

# remove extremely sparse dummy columns
# low_info = [c for c in X.columns if X[c].sum() < 20]

# X = X.drop(columns=low_info)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,       # 80% train, 20% test
    random_state=42,
    stratify=y           # VERY IMPORTANT for classification
)

# X_train = X_train.drop(columns=low_info, errors='ignore')
# X_test = X_test.drop(columns=low_info, errors='ignore')

model = RandomForestClassifier(
    n_estimators=800,
    max_features="sqrt",
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
```

```python
y.value_counts(normalize=True)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test, y_pred)
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier

model = HistGradientBoostingClassifier()
model.fit(X_train, y_train)

print("Accuracy:", accuracy_score(y_test, model.predict(X_test)))
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier

model = HistGradientBoostingClassifier(max_iter=200)
model.fit(X_train, y_train)

from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, model.predict(X_test)))
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier

model = HistGradientBoostingClassifier(
    max_iter=300,
    learning_rate=0.05,
    max_depth=6,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

print("Accuracy:", accuracy_score(y_test, model.predict(X_test)))
```
